```sql
create catalog if not exists ktproject
managed location 'abfss://layers@azdbsktsotrageacount.dfs.core.windows.net'
comment 'catalog for ktproject'
```

```sql
create schema if not exists ktproject.delta_lake
managed location 'abfss://layers@azdbsktsotrageacount.dfs.core.windows.net/delta_lake'
comment 'schema for delta lake'
```

##Create detla lake table 

```sql
create table if not exists ktproject.delta_lake.companies
(
    company_name String,
    founded_year Date,
    country String
);
```

##Insert data in Delta_lake

```sql
insert into ktproject.delta_lake.companies
values ('Microsoft', '1975-04-04', 'USA');
```

```sql
insert into ktproject.delta_lake.companies
values 
('Google', '1998-09-04', 'USA'),
('exact_science', '2014-01-01', 'USA'),
('UW', '1855-01-01', 'USA');
```

##Time_ travel and History

```sql
select * from ktproject.delta_lake.companies
version as of 2;
```
